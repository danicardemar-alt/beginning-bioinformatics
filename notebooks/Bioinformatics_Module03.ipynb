The assignment operator `=` is used to assign a value to a variable. Here are three examples:

1.  **Simple Assignment**: Assigning a direct value to a variable.
2.  **Assignment with an Expression**: Assigning the result of an expression to a variable.
3.  **Reassignment**: Changing the value of an already assigned variable.

````python
# Example 1: Simple Assignment
my_number = 10
print(f"my_number: {my_number}")

# Example 2: Assignment with an Expression
sum_result = 5 + 7
print(f"sum_result: {sum_result}")

# Example 3: Reassignment
my_variable = "Initial Value"
print(f"Initial my_variable: {my_variable}")
my_variable = "New Value"
print(f"Reassigned my_variable: {my_variable}")

# You can also use assignment operators like +=, -=, *=, /=, etc.
count = 0
count += 1 # Equivalent to: count = count + 1
print(f"Count after +=: {count}")
````

#<font color=blue>**P23.**</font> SeqUtils.gc_fraction( ) – a function to compute GC content

####**TASK P23.1** Modify the script from TASK P22.1 so that it prints the GC content for each sequence in the file

####**TASK P23.2** <font color=orange> **Complete Rosalind #9  Computing GC Content**</font>

````python
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction

highest_gc_content = -1.0
highest_gc_id = ""

# Open the FASTA file and iterate through each sequence record
for seq_record in SeqIO.parse("practice_9.txt", "fasta"):
    # Calculate GC content for the current sequence (as a fraction)
    current_gc_content = gc_fraction(seq_record.seq)

    # Check if this sequence has a higher GC content
    if current_gc_content > highest_gc_content:
        highest_gc_content = current_gc_content
        highest_gc_id = seq_record.id

# Print the ID and the highest GC content, formatted as a percentage to 6 decimal places
print(highest_gc_id)
print(f"{highest_gc_content * 100:.6f}")
````
