# **<font color=blue>P3. </font>**```=``` is the ASSIGNMENT operator (assigns values to variables)
####**TASK P3.1:** Ask Gemini, "*Give me 3 examples to help practice understanding the assignment operator.*"

````python
x=5
x
````

The assignment operator `=` is used to assign a value to a variable. Here are three examples:

1.  **Simple Assignment**: Assigning a direct value to a variable.
2.  **Assignment with an Expression**: Assigning the result of an expression to a variable.
3.  **Reassignment**: Changing the value of an already assigned variable.

````python
# Example 1: Simple Assignment
my_number = 10
print(f"my_number: {my_number}")

# Example 2: Assignment with an Expression
sum_result = 5 + 7
print(f"sum_result: {sum_result}")

# Example 3: Reassignment
my_variable = "Initial Value"
print(f"Initial my_variable: {my_variable}")
my_variable = "New Value"
print(f"Reassigned my_variable: {my_variable}")

# You can also use assignment operators like +=, -=, *=, /=, etc.
count = 0
count += 1 # Equivalent to: count = count + 1
print(f"Count after +=: {count}")
````

````python
my_string = "HelloWorldCharlesDarwinAlfred R. Wallace"

# Using find() to get the index of a character
index_of_W = my_string.find('W')
print(f"The first 'W' is at index: {index_of_W}")

# Using find() to get the index of a substring
index_of_Wallace = my_string.find('Wallace')
print(f"'Wallace' starts at index: {index_of_Wallace}")

# If the character/substring is not found, find() returns -1
index_of_Z = my_string.find('Z')
print(f"'Z' is at index: {index_of_Z} (not found)")

# index() works similarly, but raises a ValueError if not found
# index_of_Z_error = my_string.index('Z') # This would raise an error

# You can also specify a starting index for the search
index_of_second_l = my_string.find('l', 3) # Start searching from index 3
print(f"The second 'l' is at index (starting from index 3): {index_of_second_l}")
````
